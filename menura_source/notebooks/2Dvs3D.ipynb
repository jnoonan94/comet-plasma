```python
idx_it = 400
path = '..'
md_3d = menura_data(path, '', idx_it, ask_scp=False)
path = '../menura'
md = menura_data(path, '', idx_it, ask_scp=False)
```

```python
field_2d = np.log10(md.dens_spec0)[0]
mid_z = int(md_3d.mp.len_z_cst/2)
field_3d = np.log10(np.mean(md_3d.dens_spec0[0, :, :, mid_z-1:mid_z], axis=2))

fig, AX = plt.subplots(1, 2, figsize=(16, 12))
for ax in AX:
    ax.set_aspect('equal')
        
p0 = AX[0].pcolormesh(md.edges_x[0], md.edges_y[0], (field_2d).T,
                           vmin = -1, vmax = 1,
                           cmap=oC.bwr_2, rasterized=True)

p1 = AX[1].pcolormesh(md_3d.edges_x[0], md_3d.edges_y[0], (field_3d).T,
                           vmin = -1, vmax = 1,
                           cmap=oC.bwr_2, rasterized=True)

#        idx_y = int((.9-1.)*md.nb_proc*md.mp.len_y_cst+self.mp.len_y_cst)
 #       ax.text(self.edges_x[-1, int(.1*self.mp.len_x_cst)], self.edges_y[-1, idx_y], label, ha='center', va='center', fontsize=20)

#posAx = ax.get_position()
#cax = self.fig2.add_axes([posAx.x1*1., posAx.y0, 0.02, .3])
#cb = self.fig2.colorbar(p0, cax=cax, orientation='vertical')
# cb.set_label(label, rotation=0, ha='left', fontsize=24)

oT.set_spines(AX)
plt.tight_layout()
plt.show()


field_2d = np.log10(md.dens_spec1)[0]
mid_z = int(md_3d.mp.len_z_cst/2)
field_3d = np.log10(np.mean(md_3d.dens_spec1[0, :, :, mid_z-1:mid_z], axis=2))

fig, AX = plt.subplots(1, 2, figsize=(16, 12))
for ax in AX:
    ax.set_aspect('equal')
        
p0 = AX[0].pcolormesh(md.edges_x[0], md.edges_y[0], (field_2d).T,
                           vmin = -4, vmax = 1,
                           cmap=oC.bwr_2, rasterized=True)

p1 = AX[1].pcolormesh(md_3d.edges_x[0], md_3d.edges_y[0], (field_3d).T,
                           vmin = -4, vmax = 1,
                           cmap=oC.bwr_2, rasterized=True)

#        idx_y = int((.9-1.)*md.nb_proc*md.mp.len_y_cst+self.mp.len_y_cst)
 #       ax.text(self.edges_x[-1, int(.1*self.mp.len_x_cst)], self.edges_y[-1, idx_y], label, ha='center', va='center', fontsize=20)

#posAx = ax.get_position()
#cax = self.fig2.add_axes([posAx.x1*1., posAx.y0, 0.02, .3])
#cb = self.fig2.colorbar(p0, cax=cax, orientation='vertical')
# cb.set_label(label, rotation=0, ha='left', fontsize=24)

oT.set_spines(AX)
plt.tight_layout()
plt.show()
```
